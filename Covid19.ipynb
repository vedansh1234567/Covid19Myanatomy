```python
# Install dependencies as needed:
# pip install kagglehub[pandas-datasets]
import kagglehub
from kagglehub import KaggleDatasetAdapter

# Set the path to the file you'd like to load
file_path = "covid_19_clean_complete.csv"

# Load the latest version
df = kagglehub.load_dataset(
  KaggleDatasetAdapter.PANDAS,
  "imdevskp/corona-virus-report",
  file_path,
  # Provide any additional arguments like
  # sql_query or pandas_kwargs. See the
  # documenation for more information:
  # https://github.com/Kaggle/kagglehub/blob/main/README.md#kaggledatasetadapterpandas
)

print("First 5 records:", df.head())
```

```python
print(df.shape)
print(df.columns)
print(df.info())
print(df.describe())
```

```python
print(df.isnull().sum())
```

```python
print("Duplicates:", df.duplicated().sum())
```

```python
import pandas as pd
```

```python
df['Date'] = pd.to_datetime(df['Date'])
```

```python
df['Year'] = df['Date'].dt.year
df['Month'] = df['Date'].dt.month
```

```python
print("Total Confirmed Cases:", df['Confirmed'].sum())
```

```python
print("Total Deaths:", df['Deaths'].sum())
print("Total Recovered:", df['Recovered'].sum())
```

```python
latest_date = df['Date'].max()

latest = df[df['Date'] == latest_date]

top_countries = latest.sort_values(
    by='Confirmed',
    ascending=False
).head(10)

print(top_countries[['Country/Region', 'Confirmed', 'Deaths', 'Recovered']])
```

```python
latest['Death Rate'] = (
    latest['Deaths'] / latest['Confirmed']
) * 100

print(
    latest[
        ['Country/Region', 'Confirmed', 'Deaths', 'Death Rate']
    ].sort_values(
        by='Death Rate',
        ascending=False
    ).head(10)
)
```

```python
import matplotlib.pyplot as plt

top10 = latest.sort_values(
    by='Confirmed',
    ascending=False
).head(10)

plt.figure(figsize=(12,6))

plt.bar(
    top10['Country/Region'],
    top10['Confirmed']
)

plt.xticks(rotation=45)
plt.xlabel("Country")
plt.ylabel("Confirmed Cases")
plt.title("Top 10 Countries by COVID-19 Confirmed Cases")

plt.show()
```

```python
plt.figure(figsize=(10,6))

plt.scatter(
    latest['Confirmed'],
    latest['Deaths']
)

plt.xlabel("Confirmed Cases")
plt.ylabel("Deaths")
plt.title("COVID-19 Confirmed Cases vs Deaths")

plt.show()
```

```python
df_sorted = df.sort_values(by=['Country/Region', 'Date'])
df_sorted['New cases'] = df_sorted.groupby('Country/Region')['Confirmed'].diff().fillna(0)

daily = df_sorted.groupby('Date')['New cases'].sum()

plt.figure(figsize=(12,6))

plt.plot(daily.index, daily.values)

plt.xlabel("Date")
plt.ylabel("New cases")
plt.title("Worldwide Daily COVID-19 Cases")

plt.show()
```

```python
df_sorted = df.sort_values(by=['Country/Region', 'Date'])
df_sorted['New deaths'] = df_sorted.groupby('Country/Region')['Deaths'].diff().fillna(0)
daily_deaths = df_sorted.groupby('Date')['New deaths'].sum()
```

```python
df_sorted = df.sort_values(by=['Country/Region', 'Date'])
df_sorted['New deaths'] = df_sorted.groupby('Country/Region')['Deaths'].diff().fillna(0)
daily_deaths = df_sorted.groupby('Date')['New deaths'].sum()

plt.figure(figsize=(12,6))

plt.plot(daily_deaths.index, daily_deaths.values, color='red')

plt.xlabel("Date")
plt.ylabel("New Deaths")
plt.title("Worldwide Daily COVID-19 Deaths")

plt.show()
```

```python
countries = ['India', 'US', 'Brazil', 'Russia']

for country in countries:

    country_data = df[
        df['Country/Region'] == country
    ]

    plt.plot(
        country_data['Date'],
        country_data['Confirmed'],
        label=country
    )

plt.xlabel("Date")
plt.ylabel("Confirmed Cases")
plt.title("COVID-19 Cases Comparison")

plt.legend()
plt.show()
```
